```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

np.random.seed(42)
tf.random.set_seed(42)

print("Libraries imported successfully!")
print("TensorFlow version:", tf.__version__)
```

```python

from google.colab import files

uploaded = files.upload()

file_name = "AXISBANK.csv"
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python

# Convert Date and Close columns
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Close"] = pd.to_numeric(df["Close"], errors="coerce")

# Keep only the required columns
df = df[["Date", "Close"]].copy()

# Remove invalid records and duplicate dates
df = df.dropna(subset=["Date", "Close"])
df = df.drop_duplicates(subset=["Date"])

# Sort from oldest to newest
df = df.sort_values("Date").reset_index(drop=True)

# Select 2,000 consecutive daily trading records
if len(df) < 2000:
    raise ValueError("Dataset contains fewer than 2,000 valid records.")

df = df.tail(2000).reset_index(drop=True)

print("Cleaned dataset shape:", df.shape)
print("First date:", df["Date"].min().date())
print("Last date:", df["Date"].max().date())
print("Missing values:\n", df.isnull().sum())

display(df.head())
display(df.tail())
```

```python

plt.figure(figsize=(14, 5))
plt.plot(df["Date"], df["Close"], label="Actual Closing Price")
plt.title("AXIS Bank Historical Closing Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price (₹)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Model parameters
WINDOW_SIZE = 60

# Chronological 80% training and 20% testing split
split_index = int(len(df) * 0.80)

train_prices = df["Close"].iloc[:split_index].values.reshape(-1, 1)

scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(train_prices)

# Transform all records using the training-fitted scaler
scaled_prices = scaler.transform(
    df["Close"].values.reshape(-1, 1)
)

print("Window size:", WINDOW_SIZE, "trading records")
print("Training records:", split_index)
print("Testing records:", len(df) - split_index)
print("Normalization completed!")
```

```python

X, y, target_indices = [], [], []

for i in range(WINDOW_SIZE, len(scaled_prices)):
    X.append(scaled_prices[i-WINDOW_SIZE:i, 0])
    y.append(scaled_prices[i, 0])
    target_indices.append(i)

X = np.array(X)
y = np.array(y)
target_indices = np.array(target_indices)

# Reshape input for LSTM: samples, time steps, features
X = X.reshape(X.shape[0], X.shape[1], 1)

# Split sequences by their target dates, not randomly
train_mask = target_indices < split_index
test_mask = target_indices >= split_index

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python

model = Sequential([
    LSTM(64, return_sequences=True,
         input_shape=(WINDOW_SIZE, 1)),
    Dropout(0.2),

    LSTM(32),
    Dropout(0.2),

    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()
```

```python

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("Model training completed!")
```

```python

plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("LSTM Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error (normalized)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Predict the test period
y_pred_scaled = model.predict(X_test, verbose=0)

# Convert to original closing-price scale
y_pred = scaler.inverse_transform(y_pred_scaled).flatten()
y_actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).flatten()

test_dates = df["Date"].iloc[target_indices[test_mask]].reset_index(drop=True)

results = pd.DataFrame({
    "Date": test_dates,
    "Actual Close": y_actual,
    "Predicted Close": y_pred
})

display(results.head(10))
```

```python

mae = mean_absolute_error(y_actual, y_pred)
rmse = np.sqrt(mean_squared_error(y_actual, y_pred))

print(f"Mean Absolute Error (MAE): ₹{mae:.2f}")
print(f"Root Mean Squared Error (RMSE): ₹{rmse:.2f}")
```

```python

plt.figure(figsize=(14, 6))

plt.plot(
    results["Date"],
    results["Actual Close"],
    label="Actual Closing Price",
    linewidth=2
)

plt.plot(
    results["Date"],
    results["Predicted Close"],
    label="LSTM Predicted Price",
    linewidth=2
)

plt.title("AXIS Bank: Actual vs Predicted Closing Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price (₹)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

print("Predictions for unseen test dates:")
display(results.tail(15).round(2))

# Plot the last 100 test records
last_n = min(100, len(results))

plt.figure(figsize=(14, 5))
plt.plot(
    results["Date"].tail(last_n),
    results["Actual Close"].tail(last_n),
    label="Actual"
)
plt.plot(
    results["Date"].tail(last_n),
    results["Predicted Close"].tail(last_n),
    label="Predicted"
)

plt.title("LSTM Predictions on Unseen Test Dates")
plt.xlabel("Date")
plt.ylabel("Closing Price (₹)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Take the latest 60 closing prices
latest_window = scaled_prices[-WINDOW_SIZE:]

# LSTM expects: samples, time steps, features
latest_window = latest_window.reshape(1, WINDOW_SIZE, 1)

# Predict the next closing price
next_scaled = model.predict(latest_window, verbose=0)

next_close = scaler.inverse_transform(next_scaled)[0, 0]

last_date = df["Date"].iloc[-1]

# Approximate next weekday; exchange holidays are not accounted for
next_weekday = last_date + pd.offsets.BDay(1)

print("Last available date:", last_date.date())
print("Last actual closing price: ₹{:.2f}".format(
    df["Close"].iloc[-1]
))
print("Approximate next trading date:", next_weekday.date())
print("Predicted next closing price: ₹{:.2f}".format(next_close))
print("\nNote: This is a model estimate, not a guaranteed market price.")
```
